# Z-Score & Empirical Rule — Essential Practice

This notebook uses **artificial normally distributed data** to demonstrate:
- Mean and standard deviation
- Z-score calculation
- Simple Z-score based outlier check
- Empirical Rule (68–95–99.7%)
- Visualization

**Note:** The Empirical Rule applies approximately to normally distributed data.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

np.random.seed(42)

# Artificial data: mean = 50, standard deviation = 10
data = np.random.normal(loc=50, scale=10, size=1000)
df = pd.DataFrame({"value": data})

df.head()
```

```python
# Mean and standard deviation
mean = df["value"].mean()
std = df["value"].std()

print(f"Mean = {mean:.2f}")
print(f"Standard Deviation = {std:.2f}")
```

## Z-Score

Formula:

**z = (x - mean) / standard deviation**

Interpretation:
- z = 0 → value is at the mean
- z = +1 → 1 SD above the mean
- z = -2 → 2 SD below the mean

```python
# Calculate Z-score for every observation
df["z_score"] = (df["value"] - mean) / std

df.head()
```

```python
# Z-score for one observation
x = df.loc[0, "value"]
z = (x - mean) / std

print(f"Value = {x:.2f}")
print(f"Z-score = {z:.2f}")
```

## Z-Score Based Extreme-Value Check

A common classroom heuristic is **|z| > 3**.

This is not a universal definition of an outlier.

```python
extreme_values = df[df["z_score"].abs() > 3]

print(f"Number of observations with |z| > 3: {len(extreme_values)}")
extreme_values.head()
```

## Empirical Rule: 68–95–99.7%

For approximately normal data:

- About **68%** lies within ±1 SD
- About **95%** lies within ±2 SD
- About **99.7%** lies within ±3 SD

```python
within_1 = ((df["z_score"] >= -1) & (df["z_score"] <= 1)).mean() * 100
within_2 = ((df["z_score"] >= -2) & (df["z_score"] <= 2)).mean() * 100
within_3 = ((df["z_score"] >= -3) & (df["z_score"] <= 3)).mean() * 100

print(f"Within ±1 SD: {within_1:.2f}%")
print(f"Within ±2 SD: {within_2:.2f}%")
print(f"Within ±3 SD: {within_3:.2f}%")
```

```python
# Visualize the normal distribution and SD boundaries
x = np.linspace(mean - 4 * std, mean + 4 * std, 1000)
y = np.exp(-0.5 * ((x - mean) / std) ** 2) / (std * np.sqrt(2 * np.pi))

plt.figure(figsize=(10, 5))
plt.plot(x, y, linewidth=2)

for k, style in [(1, ":"), (2, "--"), (3, "-.")]:
    plt.axvline(mean - k * std, linestyle=style, label=f"-{k} SD")
    plt.axvline(mean + k * std, linestyle=style, label=f"+{k} SD")

plt.axvline(mean, linestyle="-", label="Mean")
plt.title("Empirical Rule: 68–95–99.7%")
plt.xlabel("Value")
plt.ylabel("Density")
plt.legend()
plt.grid(alpha=0.2)
plt.show()
```

## Quick Example

If:

- Mean = 50
- Standard deviation = 10
- x = 70

then:

**z = (70 − 50) / 10 = 2**

So, 70 is **2 standard deviations above the mean**.

```python
x = 70
mu = 50
sigma = 10

z = (x - mu) / sigma
print(f"Z-score = {z:.2f}")
```

## Key Takeaways

1. Z-score tells us the position of a value relative to the mean in SD units.
2. The Empirical Rule is approximately **68–95–99.7%** for normal data.
3. Z-scoring does **not** make arbitrary data normally distributed.
4. `|z| > 3` is a useful classroom heuristic for unusually extreme observations.
